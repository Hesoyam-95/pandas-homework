**2. Базова вибірка даних**

```python
import pandas as pd

df = pd.read_csv("tip.csv")

display(df ["total_bill"])
display(df [["total_bill", "tip"]])
display(df.head(10))
display(df.iloc[5])

```

**3. Фільтрація даних**

```python
# Всі рахунки більше 20 доларів:

df[df["total_bill"] > 20]
bill_over_20 = df[df["total_bill"] > 20]
print("Кількість записів:", bill_over_20.shape[0])
display(bill_over_20.head(5))

# Всі відвідувачі-жінки:

fem_vis = df[df["sex"] == "Female"]
print("Всі відвідувачі - жінки:",fem_vis.shape[0])
display(fem_vis.head(5))

# Рахунки на вечерю (`time == 'Dinner'`)

dinner = df[df["time"] == "Dinner"]
print("Кількість записів:", dinner.shape[0])
display(dinner.head(5))

# Рахунки на вечерю з сумою чеку більше 30 доларів.

dinner_over_30 = df[
    (df["time"] == "Dinner") &
    (df["total_bill"] > 30)]
print("Кількість записів:", dinner_over_30.shape[0])
display(dinner_over_30.head(5))
```
